                                        String Interpolation

```python
print("This is  string {}".format("INSERTED"))  #.format method used for string interpolation
print("The {} {}   {}".format('Mango','Apple','Banana'))  #.format method used for multiple times
print("The {2} {0}   {1}".format('Mango','Apple','Banana'))#.format method used for multiple times by specified index
print("The {0} {0}   {0}".format("Mango","Apple","Banana"))#.format method used for multiple times by specified index
print("The {k} {k}   {d}".format(d='Mango',t='Apple',k='Banana'))#.format method used for multiple times by specified keyword/variables
```

Zip() function zip upto shortest length

```python
mylist1=[1,2,3,4]
mylist2=['a','b','c','d']
for item in zip(mylist1,mylist2):  #this is called tuple packing by the use of zip function
    print(item)

mylist3=[100,200,300]
for item in zip(mylist1,mylist2,mylist3):
    print(item)

print(list(zip(mylist1,mylist2,mylist3))) #can make it to list
for a,b,c in zip(mylist1,mylist2,mylist3): #can access specific index value
    print(b)
```
